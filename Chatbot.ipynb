```python
import re
import random
from datetime import datetime


class TraditionalChatbot:
    """
    A rule-based chatbot that uses pattern matching (regex) and
    keyword-based routing to generate responses. Designed for
    extensibility toward AI service integration.
    """

    def __init__(self, bot_name="CumberBot"):
        self.bot_name = bot_name
        self.conversation_history = []
        self.context = {}  # tracks conversation state for follow-ups

        # ----- response rules: list of (compiled_regex, handler) -----
        self.rules = self._build_rules()

    # ------------------------------------------------------------------
    # Rule definitions
    # ------------------------------------------------------------------
    def _build_rules(self):
        """Return an ordered list of (pattern, handler) tuples."""
        rules = [
            # Greetings
            (re.compile(r"\b(hi|hello|hey|howdy|greetings|good\s*(morning|afternoon|evening))\b", re.I),
             self._handle_greeting),

            # Farewell
            (re.compile(r"\b(bye|goodbye|see\s*you|farewell|exit|quit)\b", re.I),
             self._handle_farewell),

            # Capabilities / help
            (re.compile(r"\b(help|capabilities|what can you do|features|commands|menu)\b", re.I),
             self._handle_help),

            # Time / date — anchor so "today" only matches time-related queries
            (re.compile(r"\b(what\s*(time|date)|current\s*(time|date)|what\s*day\s*is\s*it|today'?s?\s*date)\b", re.I),
             self._handle_datetime),

            # Weather (stub — extensible)
            (re.compile(r"\b(weather|temperature|forecast)\b", re.I),
             self._handle_weather),

            # Joke
            (re.compile(r"\b(joke|funny|laugh|humor)\b", re.I),
             self._handle_joke),

            # Math — simple arithmetic
            (re.compile(r"(\d+)\s*([+\-*/])\s*(\d+)", re.I),
             self._handle_math),

            # Identity questions
            (re.compile(r"\b(who\s*are\s*you|your\s*name|what\s*are\s*you)\b", re.I),
             self._handle_identity),

            # How are you
            (re.compile(r"\b(how\s*are\s*you|how\s*do\s*you\s*do|how\s*is\s*it\s*going)\b", re.I),
             self._handle_how_are_you),

            # Thanks
            (re.compile(r"\b(thanks|thank\s*you|thx|appreciate)\b", re.I),
             self._handle_thanks),

            # Sentiment — positive
            (re.compile(r"\b(great|awesome|wonderful|fantastic|excellent|good|nice|love)\b", re.I),
             self._handle_positive_sentiment),

            # Sentiment — negative
            (re.compile(r"\b(bad|terrible|awful|horrible|hate|angry|sad|upset)\b", re.I),
             self._handle_negative_sentiment),
        ]
        return rules

    # ------------------------------------------------------------------
    # Handlers
    # ------------------------------------------------------------------
    def _handle_greeting(self, user_input, match):
        responses = [
            f"Hello! I'm {self.bot_name}. How can I help you today?",
            f"Hey there! Welcome to {self.bot_name}. Type 'help' to see what I can do.",
            f"Hi! Nice to meet you. I'm {self.bot_name}, your friendly chatbot.",
            "Greetings! What can I do for you?",
        ]
        return random.choice(responses)

    def _handle_farewell(self, user_input, match):
        responses = [
            "Goodbye! Have a great day!",
            "See you later! It was nice chatting with you.",
            "Bye! Feel free to come back anytime.",
            "Farewell! Take care!",
        ]
        return random.choice(responses)

    def _handle_help(self, user_input, match):
        return (
            f"Here are my capabilities:\n"
            f"  1. Greetings & farewells — say hi or bye\n"
            f"  2. Date & time    — ask me the current date or time\n"
            f"  3. Simple math    — e.g., '25 + 17' or '100 / 5'\n"
            f"  4. Jokes          — ask me for a joke\n"
            f"  5. Weather info   — ask about weather (stub — extensible)\n"
            f"  6. Sentiment      — I respond to your mood\n"
            f"  7. Identity       — ask 'who are you?'\n"
            f"  8. General chat   — I'll do my best!\n"
            f"\nType 'quit' or 'exit' to end the conversation."
        )

    def _handle_datetime(self, user_input, match):
        now = datetime.now()
        return (
            f"The current date is {now.strftime('%A, %B %d, %Y')} "
            f"and the time is {now.strftime('%I:%M %p')}."
        )

    def _handle_weather(self, user_input, match):
        return (
            "I don't have a live weather feed yet, but this feature is "
            "designed to connect to an external API such as Azure Cognitive "
            "Services or OpenWeatherMap. Stay tuned!"
        )

    def _handle_joke(self, user_input, match):
        jokes = [
            "Why do programmers prefer dark mode? Because light attracts bugs!",
            "There are only 10 types of people: those who understand binary and those who don't.",
            "Why was the computer cold? It left its Windows open!",
            "What's a computer's favorite snack? Microchips!",
            "Why did the developer go broke? Because he used up all his cache.",
        ]
        return random.choice(jokes)

    def _handle_math(self, user_input, match):
        try:
            a = float(match.group(1))
            op = match.group(2)
            b = float(match.group(3))
            if op == '+':
                result = a + b
            elif op == '-':
                result = a - b
            elif op == '*':
                result = a * b
            elif op == '/':
                if b == 0:
                    return "Oops! Division by zero is not allowed."
                result = a / b
            else:
                return "I couldn't parse that math expression."
            # Display as int if whole number
            if result == int(result):
                result = int(result)
            return f"The answer is {result}."
        except Exception:
            return "I had trouble with that calculation. Please try again."

    def _handle_identity(self, user_input, match):
        return (
            f"I am {self.bot_name}, a simple rule-based chatbot built with Python. "
            f"I use pattern matching and keyword routing — no large language models here! "
            f"I was created as part of an MSAI-631 course project at the "
            f"University of the Cumberlands."
        )

    def _handle_how_are_you(self, user_input, match):
        responses = [
            "I'm doing great, thanks for asking! How about you?",
            "I'm just a bot, but I'm functioning perfectly! What's on your mind?",
            "All systems operational! How can I help?",
        ]
        return random.choice(responses)

    def _handle_thanks(self, user_input, match):
        responses = [
            "You're welcome!",
            "Happy to help!",
            "Anytime! Let me know if you need anything else.",
            "No problem at all!",
        ]
        return random.choice(responses)

    def _handle_positive_sentiment(self, user_input, match):
        responses = [
            "That's wonderful to hear!",
            "Glad things are going well!",
            "Awesome! Is there anything I can help with?",
        ]
        return random.choice(responses)

    def _handle_negative_sentiment(self, user_input, match):
        responses = [
            "I'm sorry to hear that. Is there anything I can do to help?",
            "That doesn't sound great. Would you like to talk about it?",
            "I hope things get better soon! How can I assist you?",
        ]
        return random.choice(responses)

    def _handle_fallback(self, user_input):
        """Called when no rule matches — handles malformed or unknown input."""
        responses = [
            "I'm not sure I understand. Could you rephrase that?",
            "Hmm, I didn't catch that. Type 'help' to see what I can do.",
            f"Sorry, I don't have a response for that yet. "
            f"Try asking me about the time, math, or a joke!",
            "I'm a simple bot — that's beyond my abilities right now. "
            "Type 'help' for a list of things I can do.",
        ]
        return random.choice(responses)

    # ------------------------------------------------------------------
    # Core logic
    # ------------------------------------------------------------------
    def get_response(self, user_input):
        """
        Process user input through the rule chain and return a response.
        This method is the single integration point — an AI service
        could replace or augment the fallback path.
        """
        cleaned = user_input.strip()
        if not cleaned:
            return "It looks like you didn't type anything. Try again!"

        # Walk through rules in priority order
        for pattern, handler in self.rules:
            match = pattern.search(cleaned)
            if match:
                response = handler(cleaned, match)
                self._log(cleaned, response)
                return response

        # No rule matched — fallback
        response = self._handle_fallback(cleaned)
        self._log(cleaned, response)
        return response

    def _log(self, user_input, response):
        """Keep a conversation history for potential analytics."""
        self.conversation_history.append({
            "timestamp": datetime.now().isoformat(),
            "user": user_input,
            "bot": response,
        })

    # ------------------------------------------------------------------
    # Extension hook (placeholder for AI service integration)
    # ------------------------------------------------------------------
    def query_ai_service(self, user_input):
        """
        Placeholder for connecting to an external AI service such as
        Azure Cognitive Services, AWS Lex, or Google Dialogflow.
        When implemented, this would send the user_input to the service
        and return its response, enabling natural-language understanding
        beyond simple pattern matching.
        """
        raise NotImplementedError(
            "AI service integration is not yet implemented. "
            "Override this method to connect to Azure Cognitive Services."
        )


def main():
    """Interactive terminal loop."""
    bot = TraditionalChatbot()
    print("=" * 55)
    print(f"  Welcome to {bot.bot_name}!")
    print("  Type 'help' for a list of commands, 'quit' to exit.")
    print("=" * 55)
    print()

    while True:
        try:
            user_input = input("You: ").strip()
        except (EOFError, KeyboardInterrupt):
            print("\nGoodbye!")
            break

        if not user_input:
            continue

        if re.match(r"\b(quit|exit|bye|goodbye)\b", user_input, re.I):
            print(f"{bot.bot_name}: {bot.get_response(user_input)}")
            break

        response = bot.get_response(user_input)
        print(f"{bot.bot_name}: {response}")
        print()


if __name__ == "__main__":
    main()
```
